```python
from scipy.sparse import coo_matrix
from implicit.als import AlternatingLeastSquares
from sklearn.preprocessing import LabelEncoder
import pandas as pd

df = pd.DataFrame(data)

df['track_id'] = df['song_name'].str.lower().str.replace(" ", "_") + ":" + df['artist_name'].str.lower().str.replace(" ", "_")
df['score'] = 1 / df['rank']

user_encoder = LabelEncoder()
item_encoder = LabelEncoder()

df['user_idx'] = user_encoder.fit_transform(df['user_id'])
df['item_idx'] = item_encoder.fit_transform(df['track_id'])

matrix = coo_matrix((df['score'], (df['user_idx'], df['item_idx'])))

matrix_csr = matrix.tocsr()

model = AlternatingLeastSquares(factors=64, iterations=15)
model.fit(matrix_csr)

def get_cf_recommendations(user_id, N=10):
    user_idx = user_encoder.transform([user_id])[0]
    item_idxs, scores = model.recommend(user_idx, matrix_csr[user_idx], N=N)
    track_ids = item_encoder.inverse_transform(item_idxs)
    return pd.DataFrame({'track_id': track_ids, 'cf_score': scores})

user_id = "user_003"
cf_recs = get_cf_recommendations(user_id, N=5)
print(cf_recs)

```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters for synthetic data
n_users = 100
n_items = 50
n_factors = 10  # number of latent factors

# Generating random user and item factors
np.random.seed(0)
user_factors = np.random.rand(n_users, n_factors)
item_factors = np.random.rand(n_items, n_factors)

# Generating ratings matrix
ratings = np.matmul(user_factors, item_factors.T)

# Adding some noise
noise = np.random.normal(0, 0.1, ratings.shape)
ratings += noise

# Introducing some sparsity to the matrix
ratings[np.random.rand(*ratings.Shape) > 0.8] = 0

def als(ratings, n_factors, n_iterations, lambda_):
    n_users, n_items = ratings.shape
    # Initialize user and item factors with random values
    user_factors = np.random.rand(n_users, n_factors)
    item_factors = np.random.rand(n_items, n_factors)

    # ALS algorithm
    for iteration in range(n_iterations):
        # Update user factors
        for i in range(n_users):
            non_zero_items = ratings[i, :] != 0
            YTY = item_factors[non_zero_items].T.dot(item_factors[non_zero_items])
            lambda_I = np.eye(n_factors) * lambda_
            user_factors[i] = np.linalg.solve(YTY + lambda_I, ratings[i, non_zero_items].dot(item_factors[non_zero_items]))
        
        # Update item factors
        for j in range(n_items):
            non_zero_users = ratings[:, j] != 0
            XTX = user_factors[non_zero_users].T.dot(user_factors[non_zero_users])
            item_factors[j] = np.linalg.solve(XTX + lambda_I, ratings[non_zero_users, j].dot(user_factors[non_zero_users]))

    return user_factors, item_factors

# Running ALS
n_iterations = 15
lambda_ = 0.1
user_factors, item_factors = als(ratings, n_factors, n_iterations, lambda_)
predicted_ratings = np.matmul(user_factors, item_factors.T)

# Plotting the original and predicted ratings
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(ratings, aspect='auto')
plt.title("Original Ratings")
plt.xlabel("Items")
plt.ylabel("Users")
plt.colorbar()

plt.subplot(1, 2, 2)
plt.imshow(predicted_ratings, aspect='auto')
plt.title("Predicted Ratings")
plt.xlabel("Items")
plt.ylabel("Users")
plt.colorbar()

plt.show()
```
